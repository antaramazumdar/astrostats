```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import poisson


values = [17, 11, 25, 16, 27, 10, 18, 33, 33, 21, 17, 19, 21, 27, 23, 18, 21, 27, 19, 14, 19, 12, 17, 16, 17, 21, 34, 16, 19, 28, 21, 11, 15, 17, 13]
# Initialize layout
fig, ax = plt.subplots(figsize = (10, 10))
ax.hist(values, bins=8)
plt.show()
```

```python
plt.ylabel('Probability')
plt.xlabel('Number of cosmic rays')
plt.title('Probability Distribution Curve')
arr = []
rv = poisson(18)
for num in range(0, 35):
 arr.append(rv.pmf(num))
 
plt.grid(True)
plt.plot(arr, linewidth=2.0)
plt.show()
```
